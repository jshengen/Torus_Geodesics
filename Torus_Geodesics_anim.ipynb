```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

%matplotlib Qt5Agg

#define the parameters for the torus
R = 2

r = 1

#define h, the step size
h = .001


#we will let y_1 = u, y_2 = v, y_3 = u', y-4 = v'

#we will iniitialize the values 

#this is the initial u coordinate
y_1_knot = 0

#this is the initial v coordinate
y_2_knot = 0

#angle for the initial unit velocity
gamma = np.pi/np.e

#inital u' coordinate
y_3_knot = np.cos(gamma)

#initial v' coordinate
y_4_knot = np.sin(gamma)


#define functions for the parametrization of the torus
def xFunc(u,v): 
    return (R + r*np.cos(u)) * np.cos(v)

def yFunc(u,v): 
    return (R + r*np.cos(u)) * np.sin(v)

def zFunc(u,v): 
    return r * np.sin(u)

#define the derivatives of the state variables
def fy_1(y_1, y_2, y_3, y_4):
    """this is d/dt(y_1) """
    output = y_3
    return output

def fy_2(y_1, y_2, y_3, y_4):
    """this is d/dt(y_2) """
    output = y_4
    return output

def fy_3(y_1, y_2, y_3, y_4):
    """this is d/dt(y_3) """
    output = -( np.sin(y_1) * ( R + r * np.cos(y_1) ) ) / r * (y_4)**2
    return output

def fy_4(y_1, y_2, y_3, y_4):
    """this is d/dt(y_4) """
    output = ( 2 * r * np.sin(y_1) ) / (R + r * np.cos(y_1) ) * y_3 * y_4
    return output 


#RK4 for the system
def computeNext(y_1, y_2, y_3, y_4):
    #compute the k's

    k_1_y_1 = fy_1(y_1, y_2, y_3, y_4)

    k_1_y_2 = fy_2(y_1, y_2, y_3, y_4)

    k_1_y_3 = fy_3(y_1, y_2, y_3, y_4)

    k_1_y_4 = fy_4(y_1, y_2, y_3, y_4)


    a = y_1 + 1/2 * h * k_1_y_1

    b = y_2 + 1/2 * h * k_1_y_2

    c = y_3 + 1/2 * h * k_1_y_3

    d = y_4 + 1/2 * h * k_1_y_4

    k_2_y_1 = fy_1(a, b, c, d)

    k_2_y_2 = fy_2(a, b, c, d)

    k_2_y_3 = fy_3(a, b, c, d)

    k_2_y_4 = fy_4(a, b, c, d)


    a2 = y_1 + 1/2 * h * k_2_y_1

    b2 = y_2 + 1/2 * h * k_2_y_2

    c2 = y_3 + 1/2 * h * k_2_y_3

    d2 = y_4 + 1/2 * h * k_2_y_4

    k_3_y_1 = fy_1(a2, b2, c2, d2)

    k_3_y_2 = fy_2(a2, b2, c2, d2)

    k_3_y_3 = fy_3(a2, b2, c2, d2)

    k_3_y_4 = fy_4(a2, b2, c2, d2)


    a3 = y_1 + h * k_3_y_1

    b3 = y_2 + h * k_3_y_2

    c3 = y_3 + h * k_3_y_3

    d3 = y_4 + h * k_3_y_4

    k_4_y_1 = fy_1(a3, b3, c3, d3)

    k_4_y_2 = fy_2(a3, b3, c3, d3)

    k_4_y_3 = fy_3(a3, b3, c3, d3)

    k_4_y_4 = fy_4(a3, b3, c3, d3)

    #using the calculated k's, update the variables
    y_1 = y_1 + h*(1/6*k_1_y_1 + 2/6 * k_2_y_1 + 2/6 * k_3_y_1 + 1/6 * k_4_y_1)
    y_2 = y_2 + h*(1/6*k_1_y_2 + 2/6 * k_2_y_2 + 2/6 * k_3_y_2 + 1/6 * k_4_y_2)
    y_3 = y_3 + h*(1/6*k_1_y_3 + 2/6 * k_2_y_3 + 2/6 * k_3_y_3 + 1/6 * k_4_y_3)
    y_4 = y_4 + h*(1/6*k_1_y_4 + 2/6 * k_2_y_4 + 2/6 * k_3_y_4 + 1/6 * k_4_y_4)

    return y_1, y_2, y_3, y_4

def populateUV(y_1, y_2, y_3, y_4, UList , VList):
    #store the first values of u and v
    UList[0] = y_1
    VList[0] = y_2

    for l in range(1, ListLength):
        
        y_1, y_2, y_3, y_4 = computeNext(y_1, y_2, y_3, y_4)
        
        #store the updated u and v
        UList[l] = y_1
        VList[l] = y_2
    return y_1, y_2, y_3, y_4
    

#the number of time points to calculate the geodesic for
ListLength = 500

#initialize the u, v, u', v'
y_1 = y_1_knot
y_2 = y_2_knot
y_3 = y_3_knot
y_4 = y_4_knot

#create lists to store the calculated values of u and v
UList = np.zeros(ListLength)
VList = np.zeros(ListLength)

y_1, y_2, y_3, y_4 = populateUV(y_1, y_2, y_3, y_4, UList, VList)

    







#define coordinates to plot the torus with
u = np.linspace(0, 2*np.pi, 1000, endpoint=False)
v = np.linspace(0, 2*np.pi, 1000, endpoint=False)
U, V = np.meshgrid(u, v)

#initialize your plot
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

#calculate values for the torus surface
x = xFunc(U,V)
y = yFunc(U,V)
z = zFunc(U,V)

#strideMod changes the density of the wireframe
strideMod = 80

#plot the surface of the torus and a wireframe for it
surf = ax.plot_wireframe(x, y, z, rstride=strideMod, cstride=strideMod, alpha = 0.2)
surf2 = ax.plot_surface(x, y, z, alpha = 0.1, cmap = "rainbow")


#obtain the x,y,z values of the geodesic using the obtained u,v parametrization 
XList = xFunc(UList,VList) 

YList = yFunc(UList,VList) 

ZList = zFunc(UList,VList) 

#plot the geodesic
line, = ax.plot(XList, YList, ZList, color = 'r', linewidth = 2)

#format the size of the plot so that the donut doesnt look distorted
BoxSize = r + R
ax.set_box_aspect([1, 1, 1])
ax.set_xlim(-BoxSize, BoxSize)
ax.set_ylim(-BoxSize, BoxSize)
ax.set_zlim(-BoxSize, BoxSize)




def update(frame):
    global y_1, y_2, y_3, y_4

    y_1, y_2, y_3, y_4 = populateUV(y_1, y_2, y_3, y_4, UList, VList)
        
    XList = xFunc(UList,VList) 

    YList = yFunc(UList,VList) 

    ZList = zFunc(UList,VList) 

    line.set_data(XList, YList)

    line.set_3d_properties(ZList)

    return line,

ani = FuncAnimation(fig, update, interval=20, blit=False, cache_frame_data=False)

#display the plot
plt.show()
```
